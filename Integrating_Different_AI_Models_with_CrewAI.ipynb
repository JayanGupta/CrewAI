```python
import os
from crewai import Agent, Task, Crew, Process

# Set API key via env var (recommended)
os.environ["GEMINI_API_KEY"] = "apikey"

# Define agents
researcher = Agent(
    role="Researcher",
    goal="Find information about climate change.",
    backstory="Expert in environmental science research.",
    llm="gemini/gemini-1.5-flash",  # Single string identifier for LLM
    verbose=True,
    memory=True
)

writer = Agent(
    role="Writer",
    goal="Summarize the findings into a short report.",
    backstory="Skilled in writing scientific reports for general audiences.",
    llm="gemini/gemini-1.5-flash",
    verbose=True
)

# Define tasks
task1 = Task(description="Collect recent facts about climate change.", agent=researcher, expected_output="A list of researched facts.")
task2 = Task(description="Summarize the research into a short report.", agent=writer, expected_output="A short, clear report.", context=[task1])


crew = Crew(
    agents=[researcher, writer],
    tasks=[task1, task2],
    process=Process.sequential,
    memory=True,
    embedder={
        "provider": "google",
        "config": {
            "model": "models/embedding-001",
            "api_key":"apikey"  # Same key works for embeddings
        }
    }
)

crew.kickoff()
```

```python
from crewai import Agent, Task, Crew, Process, LLM

# Setup LLM
hf_llm = LLM(
    model="huggingface/together/deepseek-ai/DeepSeek-R1",
    api_key="apikey",
)

# Define Agents
researcher = Agent(
    role="Researcher",
    goal="Find educational content and important points on the given topic.",
    backstory="Expert in gathering accurate information and breaking it into clear points for learning.",
    llm=hf_llm,
    verbose=True,
    memory=True
)

seo_writer = Agent(
    role="SEO Article Writer",
    goal="Create SEO-optimized educational articles in simple language that are clear, structured, and easy for learners to understand.",
    backstory=(
        "This agent is a skilled content writer for EdTech. It understands SEO principles "
        "and the needs of students. It avoids jargon and unnecessary complexity, making "
        "educational topics simple and engaging. The agent always uses short sentences, "
        "bullet points when needed, and avoids extra punctuation. Its mission is to make "
        "learning accessible, while also improving search visibility for the EdTech platform."
    ),
    llm=hf_llm,
    verbose=True,
    memory=True
)

# Define tasks with placeholders
task1 = Task(
    description="Research educational content and key points about {topic}.",
    agent=researcher,
    expected_output="A list of clear, factual, and useful points about {topic}."
)

task2 = Task(
    description=(
        "Write an SEO-optimized article on {topic}. Use simple words, easy readability, "
        "short sentences, and bullet points for longer explanations. Avoid unnecessary punctuation."
    ),
    agent=seo_writer,
    expected_output="A clear, SEO-friendly educational article on {topic}.",
    context=[task1]
)

# Create crew
crew = Crew(
    agents=[researcher, seo_writer],
    tasks=[task1, task2],
    process=Process.sequential,
    memory=True,
    embedder={
        "provider": "huggingface",
        "config": {
            "model": "huggingface/sentence-transformers/all-MiniLM-L6-v2",
            "api_key": "apikey"
        }
    }
)

# ---- Run for multiple topics using inputs ----
topics = [
    "Artificial Intelligence in Education",
    "Benefits of Online Learning",
    "Future of Data Science Careers"
]

for t in topics:
    print(f"\n📘 Article for: {t}\n")
    result = crew.kickoff(inputs={"topic": t})
    print(result)
    print("\n" + "="*80 + "\n")
```
