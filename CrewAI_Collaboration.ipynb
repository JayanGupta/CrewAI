```python
import os
os.environ["OPENAI_API_KEY"] = "apikey"

from crewai import Agent, Task, Crew, Process

# Define agents with backstory
chef = Agent(
    role='Chef',
    goal='Prepare a delicious meal using the best ingredients.',
    verbose=True,
    backstory="The chef has years of experience in the kitchen, specializing in a variety of cuisines. He knows how to source the freshest ingredients and create the perfect balance of flavors."
)

food_critic = Agent(
    role='Food Critic',
    goal='Evaluate the quality of the meal based on taste, presentation, and creativity.',
    verbose=True,
    backstory="The food critic has an eye for detail and a refined palate, having reviewed the best restaurants around the world. They can spot culinary excellence in every dish."
)

# Define tasks
prepare_meal_task = Task(
    description='Prepare a meal using the finest ingredients.',
    agent=chef,
    expected_output='A well-prepared and tasty meal.'
)

evaluate_meal_task = Task(
    description='Evaluate the meal based on various factors like taste and presentation.',
    agent=food_critic,
    expected_output='A detailed review of the meal with ratings on various aspects.',
    context=[prepare_meal_task]  # Specifies that this task depends on prepare_meal_task
)

# Create crew
cooking_crew = Crew(
    agents=[chef, food_critic],
    tasks=[prepare_meal_task, evaluate_meal_task],
    process=Process.sequential,
    verbose=True
)

# Execute crew
cooking_crew.kickoff()
```

```python
from crewai import Agent, Task, Crew, Process

# Define agents
construction_worker_1 = Agent(
    role='Foundation Builder',
    goal='Build the foundation of the house.',
    verbose=True,
    backstory="The foundation builder has extensive experience laying solid foundations for homes."
)

construction_worker_2 = Agent(
    role='Wall Builder',
    goal='Construct the walls of the house.',
    verbose=True,
    backstory="The wall builder specializes in framing, placing windows, and ensuring the walls are straight."
)

construction_worker_3 = Agent(
    role='Roof Builder',
    goal='Install the roof of the house.',
    verbose=True,
    backstory="The roof builder ensures the roof is securely placed and waterproof."
)

# Define tasks
build_foundation_task = Task(
    description='Build the foundation of the house.',
    agent=construction_worker_1,
    expected_output='A solid foundation built for the house.'
)

build_walls_task = Task(
    description='Construct the walls of the house.',
    agent=construction_worker_2,
    expected_output='The walls of the house are up, ready for the roof.'
)

build_roof_task = Task(
    description='Install the roof of the house.',
    agent=construction_worker_3,
    expected_output='The roof is securely in place.'
)

# Create crew with hierarchical processing
construction_crew = Crew(
    agents=[construction_worker_1, construction_worker_2, construction_worker_3],
    tasks=[build_foundation_task, build_walls_task, build_roof_task],
    process=Process.hierarchical,  # The manager oversees the task flow
    manager_llm="gpt-4o",
    verbose=True
)

# Execute crew
construction_crew.kickoff()
```
